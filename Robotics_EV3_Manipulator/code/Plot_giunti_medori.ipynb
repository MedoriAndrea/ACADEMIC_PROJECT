```python
import matplotlib.pyplot as plt
import numpy as np

FILENAME = 'file_dati.txt'  

# Lettura dati
# Ogni riga: time; q1_des; q1_reale; q2_des; q2_reale; q3_des; q3_reale; q4_des; q4_reale
data = np.loadtxt(FILENAME, delimiter=';')

t = data[:,0]
q_des = data[:, 1::2]  # colonne dispari: desiderate
q_act = data[:, 2::2]  # colonne pari: reali

fig, axs = plt.subplots(4, 1, figsize=(8, 10), sharex=True)
fig.suptitle('Andamento Giunti nel Tempo', fontsize=14)

for i, ax in enumerate(axs):
    ax.plot(t, q_des[:,i],     label=f'Q{i+1} _des')
    ax.plot(t, q_act[:,i], '--',label=f'Q{i+1} _real')
    ax.set_ylabel(f'Q{i+1} [rad]')
    ax.legend(loc='upper right')
    ax.grid(True)

plt.tight_layout()
plt.show()
```
