```python
import sympy as sp
import numpy as np
from sympy import symbols, Matrix, lambdify
import dill
```

```python
# Definizione simbolica delle variabili
q1, q2, q3, q4 = sp.symbols('q1 q2 q3 q4', real=True)
q_medio1, q_medio2, q_medio3, q_medio4 = sp.symbols('q_medio1 q_medio2 q_medio3 q_medio4', real=True)
q_M1, q_M2, q_M3, q_M4 = sp.symbols('q_M1 q_M2 q_M3 q_M4', real=True)
q_m1, q_m2, q_m3, q_m4 = sp.symbols('q_m1 q_m2 q_m3 q_m4', real=True)

# Liste delle variabili
q = [q1, q2, q3, q4]
q_ = [q_medio1, q_medio2, q_medio3, q_medio4]
q_M = [q_M1, q_M2, q_M3, q_M4]
q_m = [q_m1, q_m2, q_m3, q_m4]

# Numero di giunti
n = len(q)

# Funzionale di costo basato sulla distanza dal fine corsa
def function_cost_fine_corsa(q, q_, q_M, q_m, n):
    w = 0
    for i in range(n):
        sum = (q[i] - q_[i]) / (q_M[i] - q_m[i])
        w += sum**2
    w = -1 / (2 * n) * w
    return w

# Calcolo funzionale
w = function_cost_fine_corsa(q, q_, q_M, q_m, n)

# Calcolo del gradiente simbolico
W_q = [w.diff(q_i) for q_i in q]
print(W_q)

# Conversione in funzione numerica con lambdify
W_q_numpy = sp.lambdify(q + q_ + q_M + q_m, W_q, 'numpy')

# Salvataggio su file .pkl
with open('W_q_numpy_functions_fine_corsa.pkl', 'wb') as file:
    dill.dump(W_q_numpy, file)

```
