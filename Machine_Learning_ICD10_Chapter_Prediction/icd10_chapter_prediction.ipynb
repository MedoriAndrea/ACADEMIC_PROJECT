```python
#PROGETTO_ML_FAUSTI_MEDORI_ROSINA

# ============================================================
# PROGETTO ML - ICD CHAPTER PREDICTION
# Versione finale :
# 1) Preprocessing e feature engineering
# 2) Analisi statistica preliminare descrittiva su tutto il dataset
# 3) Nested Stratified CV per model selection
# 4) Ablation study sul miglior modello con iperparametri fissi
# 5) Fit finale su training 80% e test finale su 20%
# ============================================================

import pandas as pd
import numpy as np
import time
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap

from google.colab import files
from IPython.display import display

from scipy.stats import kruskal

from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV, cross_val_score
from sklearn.preprocessing import OneHotEncoder, StandardScaler, LabelEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectFpr, chi2
from sklearn.base import clone

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from scipy.stats import chi2_contingency
from sklearn.base import BaseEstimator, TransformerMixin

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
    accuracy_score,
    balanced_accuracy_score,
    cohen_kappa_score
)

# XGBoost
try:
    from xgboost import XGBClassifier
    XGBOOST_AVAILABLE = True
except ImportError:
    XGBOOST_AVAILABLE = False
    print("XGBoost non installato. In Colab esegui: !pip install xgboost")

RANDOM_STATE = 42
P_VALUE_THRESHOLD = 0.001

USE_XGBOOST_GPU = False

# Nested CV: per non renderla troppo pesante uso 5 outer fold e 3 inner fold.
# Se vuoi massima simmetria puoi mettere INNER_SPLITS = 5.
OUTER_SPLITS = 5
INNER_SPLITS = 3


# ============================================================
# MODIFICA GRAFICI: STILE VISIVO
# ============================================================

PLOT_COLORS = {
    "blue": "#2F80ED",
    "light_blue": "#56CCF2",
    "green": "#27AE60",
    "violet": "#9B51E0",
    "deep_blue": "#1F4E79",
    "soft_blue": "#BFE9F5",
    "soft_green": "#BFE8D2",
    "soft_violet": "#D8C7F4",
    "dark_text": "#243447",
    "grid": "#DDE7F0"
}

# MODIFICA GRAFICI:

HARMONIOUS_PALETTE = [
    PLOT_COLORS["blue"],
    PLOT_COLORS["light_blue"],
    PLOT_COLORS["green"],
    PLOT_COLORS["violet"],
    PLOT_COLORS["deep_blue"]
]

CONFUSION_CMAP = LinearSegmentedColormap.from_list(
    "blue_green_violet",
    ["#F7FBFF", "#D6F3FA", "#A8E6CF", "#7FB3D5", "#6C63FF", "#4B3F72"]
)

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.edgecolor": "#B8C7D9",
    "axes.labelcolor": PLOT_COLORS["dark_text"],
    "axes.titlecolor": PLOT_COLORS["dark_text"],
    "xtick.color": PLOT_COLORS["dark_text"],
    "ytick.color": PLOT_COLORS["dark_text"],
    "font.size": 11,
    "axes.titlesize": 15,
    "axes.titleweight": "bold",
    "axes.labelsize": 12,
    "legend.fontsize": 10,
    "grid.color": PLOT_COLORS["grid"],
    "grid.linestyle": "--",
    "grid.alpha": 0.55
})


def polish_axes(ax, title=None, xlabel=None, ylabel=None, grid_axis="y"):
    """Rende titoli, etichette, griglia e bordi più puliti e leggibili."""
    if title is not None:
        ax.set_title(title, pad=16, loc="center", fontweight="bold")
    if xlabel is not None:
        ax.set_xlabel(xlabel, labelpad=10)
    if ylabel is not None:
        ax.set_ylabel(ylabel, labelpad=10)

    ax.grid(True, axis=grid_axis)
    ax.set_axisbelow(True)
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_color("#B8C7D9")
    ax.spines["bottom"].set_color("#B8C7D9")


def color_sequence(n, mode="auto", single_color=None):

    if n <= 0:
        return []

    if single_color is None:
        single_color = PLOT_COLORS["blue"]

    if mode == "single":
        return single_color

    if mode == "gradient":
        gradient_cmap = LinearSegmentedColormap.from_list(
            "harmonious_gradient",
            [
                PLOT_COLORS["soft_blue"],
                PLOT_COLORS["light_blue"],
                PLOT_COLORS["green"],
                PLOT_COLORS["violet"],
                PLOT_COLORS["deep_blue"]
            ]
        )
        return [gradient_cmap(i / max(n - 1, 1)) for i in range(n)]

    # mode == "auto"
    if n <= 5:
        return HARMONIOUS_PALETTE[:n]
    elif n <= 20:
        return single_color
    else:
        return color_sequence(n, mode="gradient")


# ============================================================
# FUNZIONI UTILI
# ============================================================

class CategoricalChi2Selector(BaseEstimator, TransformerMixin):
    """
    Selettore per feature categoriche originali.

    Per ogni colonna categorica:
    - costruisce una tabella di contingenza feature x target;
    - applica il test chi-quadro di indipendenza;
    - mantiene la feature originale se p-value < alpha.


    """

    def __init__(self, alpha=0.001):
        self.alpha = alpha

    def fit(self, X, y):
        X = np.asarray(X)
        y = np.asarray(y)

        self.pvalues_ = []
        self.scores_ = []

        for j in range(X.shape[1]):
            feature_values = X[:, j]

            contingency = pd.crosstab(feature_values, y)

            try:
                chi2_stat, p_value, dof, expected = chi2_contingency(contingency)
            except Exception:
                chi2_stat = 0.0
                p_value = 1.0

            if np.isnan(p_value):
                p_value = 1.0

            self.scores_.append(chi2_stat)
            self.pvalues_.append(p_value)

        self.scores_ = np.array(self.scores_)
        self.pvalues_ = np.array(self.pvalues_)

        self.support_ = self.pvalues_ < self.alpha

        return self

    def transform(self, X):
        X = np.asarray(X)
        return X[:, self.support_]

    def get_support(self):
        return self.support_

def make_ohe():
    """Compatibilità tra versioni diverse di scikit-learn."""
    try:
        return OneHotEncoder(handle_unknown="ignore", sparse_output=False)
    except TypeError:
        return OneHotEncoder(handle_unknown="ignore", sparse=False)


def get_feature_types(X_df):
    """
    Restituisce colonne numeriche e categoriche di un DataFrame.

    MODIFICA: le feature numeriche/binare codificate come 0/1 vengono trattate
    come categoriche, non come numeriche. Questo vale per colonne tipo flag,
    missing indicator e variabili booleane salvate come interi.
    """
    categorical_cols = X_df.select_dtypes(include=["object", "category", "bool"]).columns.tolist()

    numeric_candidate_cols = X_df.select_dtypes(exclude=["object", "category", "bool"]).columns.tolist()
    binary_numeric_cols = []
    true_numeric_cols = []

    for col in numeric_candidate_cols:
        non_missing_values = pd.Series(X_df[col]).dropna().unique()

        # Se i valori osservati sono solo 0 e/o 1, la colonna è una categorica binaria.
        if len(non_missing_values) > 0 and set(non_missing_values).issubset({0, 1, 0.0, 1.0}):
            binary_numeric_cols.append(col)
        else:
            true_numeric_cols.append(col)

    categorical_cols = categorical_cols + binary_numeric_cols
    numeric_cols = true_numeric_cols

    return numeric_cols, categorical_cols


def kruskal_score_func(X, y):
    """
    Score function per SelectFpr sulle feature numeriche.

    Per ogni feature numerica:
    - divide i valori per classe;
    - applica Kruskal-Wallis;
    - restituisce score = -log10(p-value) e p-value.
    """
    X = np.asarray(X)
    y = np.asarray(y)

    scores = []
    pvalues = []
    classes = np.unique(y)

    for j in range(X.shape[1]):
        groups = [X[y == c, j] for c in classes]

        try:
            stat, p = kruskal(*groups)
        except Exception:
            p = 1.0

        if np.isnan(p):
            p = 1.0

        score = -np.log10(p + 1e-300)
        scores.append(score)
        pvalues.append(p)

    return np.array(scores), np.array(pvalues)


def make_preprocessor(numeric_cols, categorical_cols, scale_numeric=False):
    """
    Preprocessing base:
    - numeriche: imputazione media + scaling solo se richiesto
    - categoriche: imputazione moda + one-hot encoding
    """
    transformers = []

    if len(numeric_cols) > 0:
        if scale_numeric:
            numeric_transformer = Pipeline(steps=[
                ("imputer", SimpleImputer(strategy="mean")),
                ("scaler", StandardScaler())
            ])
        else:
            numeric_transformer = Pipeline(steps=[
                ("imputer", SimpleImputer(strategy="mean"))
            ])
        transformers.append(("num", numeric_transformer, numeric_cols))

    if len(categorical_cols) > 0:
        categorical_transformer = Pipeline(steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", make_ohe())
        ])
        transformers.append(("cat", categorical_transformer, categorical_cols))

    preprocessor = ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )

    return preprocessor


def make_filter_preprocessor(numeric_cols, categorical_cols, scale_numeric=False):
    """
    Preprocessing con feature selection statistica p < 0.001.

    Numeriche:
    - imputazione
    - Kruskal-Wallis
    - SelectFpr(alpha=0.001)
    - scaling solo se richiesto

    Categoriche:
    - imputazione
    - chi-square sulle feature categoriche originali
    - one-hot encoding solo dopo la selezione

    La selezione è dentro la Pipeline, quindi viene rifatta dentro ogni fold.
    """
    transformers = []

    if len(numeric_cols) > 0:
        if scale_numeric:
            numeric_transformer = Pipeline(steps=[
                ("imputer", SimpleImputer(strategy="mean")),
                ("selector", SelectFpr(score_func=kruskal_score_func, alpha=P_VALUE_THRESHOLD)),
                ("scaler", StandardScaler())
            ])
        else:
            numeric_transformer = Pipeline(steps=[
                ("imputer", SimpleImputer(strategy="mean")),
                ("selector", SelectFpr(score_func=kruskal_score_func, alpha=P_VALUE_THRESHOLD))
            ])
        transformers.append(("num", numeric_transformer, numeric_cols))

    if len(categorical_cols) > 0:
        categorical_transformer = Pipeline(steps=[
            ("imputer", SimpleImputer(strategy="most_frequent")),

            # MODIFICA: il chi-quadro viene fatto sulle feature categoriche originali,
            # non sulle colonne one-hot.
            ("selector", CategoricalChi2Selector(alpha=P_VALUE_THRESHOLD)),

            # MODIFICA: il one-hot viene applicato solo dopo aver selezionato
            # le feature categoriche originali significative.
            ("onehot", make_ohe())
        ])

        # MODIFICA ERRORE: questo append deve stare dentro l'if.
        # Se non ci sono categoriche, categorical_transformer non viene creato.
        transformers.append(("cat", categorical_transformer, categorical_cols))

    preprocessor = ColumnTransformer(
        transformers=transformers,
        remainder="drop"
    )

    return preprocessor

def build_pipelines(numeric_cols, categorical_cols, n_classes):
    """Costruisce le pipeline con le colonne fornite."""
    pipelines = {}

    # Dummy
    pipelines["Dummy"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_preprocessor(numeric_cols, categorical_cols, scale_numeric=False)),
            ("model", DummyClassifier(random_state=RANDOM_STATE))
        ]),
        "params": {
            "model__strategy": ["most_frequent"]
        }
    }

    # Logistic Regression base
    pipelines["LR_base"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_preprocessor(numeric_cols, categorical_cols, scale_numeric=True)),
            ("model", LogisticRegression(
                penalty="elasticnet",
                solver="saga",
                max_iter=5000,
                random_state=RANDOM_STATE
            ))
        ]),
        "params": {
            "model__C": [0.01, 0.1],
            "model__l1_ratio": [0.2, 0.5],
            "model__class_weight": [None, "balanced"]
        }
    }

    # Logistic Regression + filter
    pipelines["LR_filter"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_filter_preprocessor(numeric_cols, categorical_cols, scale_numeric=True)),
            ("model", LogisticRegression(
                penalty="elasticnet",
                solver="saga",
                max_iter=5000,
                random_state=RANDOM_STATE
            ))
        ]),
        "params": {
            "model__C": [0.01, 0.1],
            "model__l1_ratio": [0.2, 0.5],
            "model__class_weight": [None, "balanced"]
        }
    }

    # Logistic Regression + LDA
    pipelines["LR_LDA"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_preprocessor(numeric_cols, categorical_cols, scale_numeric=True)),
            ("lda", LinearDiscriminantAnalysis()),
            ("post_lda_scaler", StandardScaler()),
            ("model", LogisticRegression(
                penalty="elasticnet",
                solver="saga",
                max_iter=5000,
                random_state=RANDOM_STATE
            ))
        ]),
        "params": {
            "model__C": [0.01, 0.1],
            "model__l1_ratio": [0.2, 0.5],
            "model__class_weight": [None, "balanced"]
        }
    }

    # Logistic Regression + filter + LDA
    pipelines["LR_filter_LDA"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_filter_preprocessor(numeric_cols, categorical_cols, scale_numeric=True)),
            ("lda", LinearDiscriminantAnalysis()),
            ("post_lda_scaler", StandardScaler()),
            ("model", LogisticRegression(
                penalty="elasticnet",
                solver="saga",
                max_iter=5000,
                random_state=RANDOM_STATE
            ))
        ]),
        "params": {
            "model__C": [0.01, 0.1],
            "model__l1_ratio": [0.2, 0.5],
            "model__class_weight": [None, "balanced"]
        }
    }

    # SVM RBF base
    pipelines["SVM_RBF_base"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_preprocessor(numeric_cols, categorical_cols, scale_numeric=True)),
            ("model", SVC(kernel="rbf", random_state=RANDOM_STATE))
        ]),
        "params": {
            "model__C": [1, 10],
            "model__gamma": [0.01, 0.1],
            "model__class_weight": [None, "balanced"]
        }
    }

    # SVM RBF + filter
    pipelines["SVM_RBF_filter"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_filter_preprocessor(numeric_cols, categorical_cols, scale_numeric=True)),
            ("model", SVC(kernel="rbf", random_state=RANDOM_STATE))
        ]),
        "params": {
            "model__C": [1, 10],
            "model__gamma": [0.01, 0.1],
            "model__class_weight": [None, "balanced"]
        }
    }

    # SVM RBF + LDA
    pipelines["SVM_RBF_LDA"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_preprocessor(numeric_cols, categorical_cols, scale_numeric=True)),
            ("lda", LinearDiscriminantAnalysis()),
            ("post_lda_scaler", StandardScaler()),
            ("model", SVC(kernel="rbf", random_state=RANDOM_STATE))
        ]),
        "params": {
            "model__C": [1, 10],
            "model__gamma": [0.01, 0.1],
            "model__class_weight": [None, "balanced"]
        }
    }

    # SVM RBF + filter + LDA
    pipelines["SVM_RBF_filter_LDA"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_filter_preprocessor(numeric_cols, categorical_cols, scale_numeric=True)),
            ("lda", LinearDiscriminantAnalysis()),
            ("post_lda_scaler", StandardScaler()),
            ("model", SVC(kernel="rbf", random_state=RANDOM_STATE))
        ]),
        "params": {
            "model__C": [1, 10],
            "model__gamma": [0.01, 0.1],
            "model__class_weight": [None, "balanced"]
        }
    }

    # Decision Tree
    pipelines["DecisionTree"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_preprocessor(numeric_cols, categorical_cols, scale_numeric=False)),
            ("model", DecisionTreeClassifier(
                random_state=RANDOM_STATE,
                min_samples_leaf=10
            ))
        ]),
        "params": {
            "model__criterion": ["gini", "entropy"],
            "model__max_depth": [5, 8],
            "model__class_weight": [None, "balanced"]
        }
    }

    # Random Forest
    pipelines["RandomForest"] = {
        "pipeline": Pipeline(steps=[
            ("preprocessor", make_preprocessor(numeric_cols, categorical_cols, scale_numeric=False)),
            ("model", RandomForestClassifier(
                random_state=RANDOM_STATE,
                n_jobs=-1,
                min_samples_leaf=5,
                n_estimators=500
            ))
        ]),
        "params": {
            "model__max_depth": [10, 20],
            "model__max_features": ["sqrt", 0.5],
            "model__class_weight": [None, "balanced"]
        }
    }

    # XGBoost
    if XGBOOST_AVAILABLE:
        xgb_kwargs = {
            "objective": "multi:softprob",
            "num_class": n_classes,
            "eval_metric": "mlogloss",
            "random_state": RANDOM_STATE,
            "n_jobs": -1,
            "n_estimators": 300,
            "colsample_bytree": 1.0
        }

        if USE_XGBOOST_GPU:
            xgb_kwargs.update({
                "tree_method": "hist",
                "device": "cuda"
            })

        pipelines["XGBoost"] = {
            "pipeline": Pipeline(steps=[
                ("preprocessor", make_preprocessor(numeric_cols, categorical_cols, scale_numeric=False)),
                ("model", XGBClassifier(**xgb_kwargs))
            ]),
            "params": {
                "model__max_depth": [3, 4],
                "model__learning_rate": [0.03, 0.1],
                "model__subsample": [0.8, 1.0]
            }
        }

    return pipelines


def safe_display_df(df, n=20):
    """Mostra le prime n righe con display se disponibile."""
    display(df.head(n))


# ============================================================
# 1. CARICAMENTO FILE
# ============================================================

print("Carica i 4 file: Demographic.csv, BloodTest.csv, Triage.csv, ICD_chapter.csv")
uploaded = files.upload()

demographic = pd.read_csv("Demographic.csv")
blood = pd.read_csv("BloodTest.csv")
triage = pd.read_csv("Triage.csv")
icd = pd.read_csv("ICD_chapter.csv")


# ============================================================
# 2. DATA AUDIT INIZIALE
# ============================================================

def audit_file(name, df):
    print(f"\n--- {name} ---")
    print("Shape:", df.shape)
    print("Duplicati hadm_id:", df["hadm_id"].duplicated().sum())
    print("Missing hadm_id:", df["hadm_id"].isna().sum())
    print("Prime colonne:", df.columns.tolist()[:10])

for name, df in {
    "Demographic": demographic,
    "BloodTest": blood,
    "Triage": triage,
    "ICD": icd
}.items():
    audit_file(name, df)

sets = {
    "Demographic": set(demographic["hadm_id"]),
    "BloodTest": set(blood["hadm_id"]),
    "Triage": set(triage["hadm_id"]),
    "ICD": set(icd["hadm_id"])
}

common_ids = set.intersection(*sets.values())

print("\n--- Controllo hadm_id ---")
print("hadm_id comuni a tutti i file:", len(common_ids))

for name, ids in sets.items():
    print(f"{name}: totale={len(ids)}, non comuni={len(ids - common_ids)}")


# ============================================================
# 3. DEMOGRAPHIC: durata ricovero e death flag
# ============================================================

demographic_processed = demographic.copy()

demographic_processed["admittime"] = pd.to_datetime(demographic_processed["admittime"], errors="coerce")
demographic_processed["dischtime"] = pd.to_datetime(demographic_processed["dischtime"], errors="coerce")
demographic_processed["deathtime"] = pd.to_datetime(demographic_processed["deathtime"], errors="coerce")

demographic_processed["length_of_stay_hours"] = (
    demographic_processed["dischtime"] - demographic_processed["admittime"]
).dt.total_seconds() / 3600

demographic_processed["death_flag"] = demographic_processed["deathtime"].notna().astype(int)

n_negative_los = (demographic_processed["length_of_stay_hours"] < 0).sum()
print("\nDurate ricovero negative:", n_negative_los)

demographic_processed.loc[
    demographic_processed["length_of_stay_hours"] < 0,
    "length_of_stay_hours"
] = np.nan

demographic_processed = demographic_processed.drop(
    columns=["admittime", "dischtime", "deathtime"]
)


# ============================================================
# 4. BLOOD TEST: parsing liste e statistiche
# ============================================================

def parse_blood_values(value):
    """Converte una cella tipo '12.3;13.1;14.0' in lista di float."""
    if pd.isna(value):
        return []

    value = str(value).strip()
    if value == "":
        return []

    values = []
    for part in value.split(";"):
        part = part.strip()
        if part == "":
            continue
        try:
            values.append(float(part))
        except ValueError:
            continue

    return values

blood_features = [col for col in blood.columns if col != "hadm_id"]
blood_processed = blood[["hadm_id"]].copy()

for col in blood_features:
    parsed = blood[col].apply(parse_blood_values)

    blood_processed[f"{col}_missing"] = parsed.apply(lambda x: int(len(x) == 0))
    blood_processed[f"{col}_mean"] = parsed.apply(lambda x: np.mean(x) if len(x) > 0 else np.nan)
    blood_processed[f"{col}_first"] = parsed.apply(lambda x: x[0] if len(x) > 0 else np.nan)
    blood_processed[f"{col}_last"] = parsed.apply(lambda x: x[-1] if len(x) > 0 else np.nan)
    blood_processed[f"{col}_min"] = parsed.apply(lambda x: np.min(x) if len(x) > 0 else np.nan)
    blood_processed[f"{col}_max"] = parsed.apply(lambda x: np.max(x) if len(x) > 0 else np.nan)
    blood_processed[f"{col}_trend"] = parsed.apply(lambda x: x[-1] - x[0] if len(x) > 0 else np.nan)
    blood_processed[f"{col}_std"] = parsed.apply(lambda x: np.std(x, ddof=0) if len(x) > 1 else (0 if len(x) == 1 else np.nan))
    blood_processed[f"{col}_n_measurements"] = parsed.apply(len)


# ============================================================
# 5. TRIAGE: valori non validi
# ============================================================

triage_processed = triage.copy()
triage_features = [col for col in triage_processed.columns if col != "hadm_id"]

vital_signs = ["temperature", "heartrate", "resprate", "o2sat", "sbp", "dbp"]

for col in vital_signs:
    if col in triage_processed.columns:
        triage_processed.loc[triage_processed[col] <= 0, col] = np.nan

if "o2sat" in triage_processed.columns:
    triage_processed.loc[(triage_processed["o2sat"] < 30) | (triage_processed["o2sat"] > 100), "o2sat"] = np.nan

if "temperature" in triage_processed.columns:
    triage_processed.loc[(triage_processed["temperature"] < 80) | (triage_processed["temperature"] > 115), "temperature"] = np.nan

if "heartrate" in triage_processed.columns:
    triage_processed.loc[(triage_processed["heartrate"] < 20) | (triage_processed["heartrate"] > 250), "heartrate"] = np.nan

if "resprate" in triage_processed.columns:
    triage_processed.loc[(triage_processed["resprate"] < 3) | (triage_processed["resprate"] > 80), "resprate"] = np.nan

if "sbp" in triage_processed.columns:
    triage_processed.loc[(triage_processed["sbp"] < 40) | (triage_processed["sbp"] > 300), "sbp"] = np.nan

if "dbp" in triage_processed.columns:
    triage_processed.loc[(triage_processed["dbp"] < 20) | (triage_processed["dbp"] > 200), "dbp"] = np.nan


# ============================================================
# 6. TARGET A 3 CLASSI
# ============================================================

icd_processed = icd.copy()

print("\nColonne ICD:", icd_processed.columns.tolist())
print(icd_processed.head())

target_col_icd = "ICD chapter"

if target_col_icd not in icd_processed.columns:
    raise ValueError(
        f"Colonna target '{target_col_icd}' non trovata. "
        f"Colonne disponibili: {icd_processed.columns.tolist()}"
    )


def map_icd_to_3_classes(value):
    value = str(value).strip()

    if value == "V":
        return "V"
    elif value in ["VI", "X", "XIII", "VI-X-XIII"]:
        return "VI-X-XIII"
    elif value == "XI":
        return "XI"
    else:
        return np.nan


icd_processed["target_3class"] = icd_processed[target_col_icd].apply(map_icd_to_3_classes)

print("\nDistribuzione target 3 classi:")
print(icd_processed["target_3class"].value_counts(dropna=False))
print(icd_processed["target_3class"].value_counts(normalize=True, dropna=False) * 100)

icd_processed = icd_processed[["hadm_id", "target_3class"]]


# ============================================================
# 7. MERGE FINALE
# ============================================================

dataset_finale = demographic_processed.merge(blood_processed, on="hadm_id", how="inner")
dataset_finale = dataset_finale.merge(triage_processed, on="hadm_id", how="inner")
dataset_finale = dataset_finale.merge(icd_processed, on="hadm_id", how="inner")

dataset_finale = dataset_finale.dropna(subset=["target_3class"])

print("\n--- Dataset finale ---")
print("Shape finale:", dataset_finale.shape)
print("Distribuzione classi:")
print(dataset_finale["target_3class"].value_counts())
print(dataset_finale["target_3class"].value_counts(normalize=True) * 100)


# ============================================================
# 8. CREAZIONE GRUPPI DI FEATURE PER ABLATION
# ============================================================

demographic_cols = [col for col in demographic_processed.columns if col != "hadm_id"]
blood_cols = [col for col in blood_processed.columns if col != "hadm_id"]
triage_cols = [col for col in triage_processed.columns if col != "hadm_id"]

feature_groups = {
    "demographic": demographic_cols,
    "blood": blood_cols,
    "triage": triage_cols,
    "demographic_blood": demographic_cols + blood_cols,
    "demographic_triage": demographic_cols + triage_cols,
    "blood_triage": blood_cols + triage_cols,
    "all": demographic_cols + blood_cols + triage_cols
}

print("\nNumero feature per gruppo:")
for group_name, cols in feature_groups.items():
    print(group_name, ":", len(cols))


# ============================================================
# 9. SALVATAGGIO DATASET FINALE
# ============================================================

output_file = "dataset_preprocessed.csv"
dataset_finale.to_csv(output_file, index=False, sep=";", decimal=",")
print("\nDataset finale creato correttamente.")
print("File salvato come:", output_file)
files.download(output_file)


# ============================================================
# 10. ANALISI STATISTICA PRELIMINARE DESCRITTIVA
#     Sull'intero dataset, solo per interpretazione/ranking.
#     La selezione predittiva viene comunque rifatta dentro la CV.
# ============================================================

print("\n" + "=" * 90)
print("ANALISI STATISTICA PRELIMINARE")
print("=" * 90)

# Istogramma/barplot sbilanciamento classi
class_counts = dataset_finale["target_3class"].value_counts()
class_percentages = dataset_finale["target_3class"].value_counts(normalize=True) * 100

class_distribution_df = pd.DataFrame({
    "class": class_counts.index,
    "count": class_counts.values,
    "percentage": class_percentages.loc[class_counts.index].values
})

print("\nDistribuzione classi iniziale:")
display(class_distribution_df)

fig, ax = plt.subplots(figsize=(7.5, 5.3))

# MODIFICA GRAFICO: barre colorate con palette armoniosa.
bars = ax.bar(
    class_distribution_df["class"],
    class_distribution_df["count"],
    color=color_sequence(len(class_distribution_df)),
    edgecolor="white",
    linewidth=1.4
)

polish_axes(
    ax,
    title="Distribuzione delle classi ICD",
    xlabel="Classe ICD",
    ylabel="Numero campioni",
    grid_axis="y"
)

# MODIFICA GRAFICO: percentuali centrate sopra ogni barra.
y_offset = class_distribution_df["count"].max() * 0.015
for bar, percentage in zip(bars, class_distribution_df["percentage"]):
    height = bar.get_height()
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        height + y_offset,
        f"{percentage:.1f}%",
        ha="center",
        va="bottom",
        fontsize=11,
        fontweight="bold",
        color=PLOT_COLORS["dark_text"]
    )

ax.margins(y=0.12)
plt.tight_layout()
plt.show()

# Feature e target per analisi statistica
X_stats = dataset_finale.drop(columns=["hadm_id", "target_3class"])
y_stats = dataset_finale["target_3class"]

label_encoder_stats = LabelEncoder()
y_stats_encoded = label_encoder_stats.fit_transform(y_stats)

stats_numeric_cols, stats_categorical_cols = get_feature_types(X_stats)

print("\nNumero feature numeriche per analisi statistica:", len(stats_numeric_cols))
print("Numero feature categoriche per analisi statistica:", len(stats_categorical_cols))

# ----------------------------
# 10.1 Kruskal-Wallis preliminare su numeriche
# ----------------------------

kruskal_rows = []

for col in stats_numeric_cols:
    groups = []
    for cls in label_encoder_stats.classes_:
        values = dataset_finale.loc[dataset_finale["target_3class"] == cls, col].dropna().values
        if len(values) > 0:
            groups.append(values)

    if len(groups) < 2:
        stat = np.nan
        p = np.nan
    else:
        try:
            stat, p = kruskal(*groups)
        except Exception:
            stat = np.nan
            p = np.nan

    kruskal_rows.append({
        "feature": col,
        "test": "Kruskal-Wallis",
        "statistic": stat,
        "p_value": p,
        "score_-log10_p": -np.log10(p + 1e-300) if pd.notna(p) else np.nan,
        "significant_p005": bool(pd.notna(p) and p < P_VALUE_THRESHOLD)
    })

kruskal_results_df = pd.DataFrame(kruskal_rows).sort_values(by="p_value", ascending=True)

print("\nTop 20 feature numeriche secondo Kruskal-Wallis:")
display(kruskal_results_df.head(20))

print("\nNumero feature numeriche significative p < 0.001:", kruskal_results_df["significant_p005"].sum())

# Barplot top Kruskal
kruskal_top_plot = kruskal_results_df.dropna(subset=["score_-log10_p"]).head(15).sort_values("score_-log10_p")

fig, ax = plt.subplots(figsize=(10.5, 7.5))

# MODIFICA GRAFICO: barre orizzontali colorate e più leggibili.
ax.barh(
    kruskal_top_plot["feature"],
    kruskal_top_plot["score_-log10_p"],
    color=color_sequence(len(kruskal_top_plot)),
    edgecolor="white",
    linewidth=1.1
)

polish_axes(
    ax,
    title="Top feature numeriche - Kruskal-Wallis",
    xlabel="-log10(p-value)",
    ylabel="Feature numerica",
    grid_axis="x"
)

plt.tight_layout()
plt.show()

# Boxplot per prime feature numeriche significative
boxplot_features = kruskal_results_df[kruskal_results_df["significant_p005"]]["feature"].head(5).tolist()

for feature in boxplot_features:
    fig, ax = plt.subplots(figsize=(7.5, 5.3))

    # MODIFICA GRAFICO: boxplot con colori tenui e testi centrati.
    dataset_finale.boxplot(
        column=feature,
        by="target_3class",
        ax=ax,
        grid=False,
        patch_artist=True,
        boxprops=dict(facecolor=PLOT_COLORS["light_blue"], color=PLOT_COLORS["blue"], linewidth=1.4),
        medianprops=dict(color=PLOT_COLORS["violet"], linewidth=2),
        whiskerprops=dict(color=PLOT_COLORS["blue"], linewidth=1.2),
        capprops=dict(color=PLOT_COLORS["blue"], linewidth=1.2),
        flierprops=dict(marker="o", markerfacecolor=PLOT_COLORS["green"], markeredgecolor=PLOT_COLORS["violet"], markersize=4, alpha=0.65)
    )

    plt.suptitle("")
    polish_axes(
        ax,
        title=f"Boxplot di {feature} per classe ICD",
        xlabel="Classe ICD",
        ylabel=feature,
        grid_axis="y"
    )
    plt.tight_layout()
    plt.show()

# Istogrammi per prime feature numeriche significative
for feature in boxplot_features:
    fig, ax = plt.subplots(figsize=(8.5, 5.5))

    # MODIFICA GRAFICO:
    for i, cls in enumerate(label_encoder_stats.classes_):
        values = dataset_finale.loc[dataset_finale["target_3class"] == cls, feature].dropna()
        ax.hist(
            values,
            bins=30,
            alpha=0.58,
            label=cls,
            color=HARMONIOUS_PALETTE[i % len(HARMONIOUS_PALETTE)],
            edgecolor="white",
            linewidth=0.4
        )

    polish_axes(
        ax,
        title=f"Istogramma di {feature} per classe ICD",
        xlabel=feature,
        ylabel="Frequenza",
        grid_axis="y"
    )
    ax.legend(title="Classe ICD", frameon=True, facecolor="white", edgecolor="#DDE7F0")
    plt.tight_layout()
    plt.show()

# ----------------------------
# 10.2 Chi-square preliminare su categoriche ORIGINALI
# ----------------------------

if len(stats_categorical_cols) > 0:

    cat_imputer = SimpleImputer(strategy="most_frequent")

    X_cat_imp = pd.DataFrame(
        cat_imputer.fit_transform(X_stats[stats_categorical_cols]),
        columns=stats_categorical_cols
    )

    chi2_rows = []

    for col in stats_categorical_cols:
        contingency = pd.crosstab(X_cat_imp[col], y_stats)

        try:
            chi2_stat, p_value, dof, expected = chi2_contingency(contingency)
        except Exception:
            chi2_stat = np.nan
            p_value = np.nan

        chi2_rows.append({
            "feature": col,
            "test": "Chi-square",
            "chi2_statistic": chi2_stat,
            "p_value": p_value,
            "score_-log10_p": -np.log10(p_value + 1e-300) if pd.notna(p_value) else np.nan,
            "significant_p005": bool(pd.notna(p_value) and p_value < P_VALUE_THRESHOLD)
        })

    chi2_results_df = pd.DataFrame(chi2_rows).sort_values(
        by="p_value",
        ascending=True
    )

    print("\nParametri usati per il chi-square preliminare:")
    print("test = scipy.stats.chi2_contingency")
    print("input = feature categoriche originali dopo imputazione most_frequent")
    print("soglia significatività = p <", P_VALUE_THRESHOLD)

    print("\nFeature categoriche originali secondo Chi-square:")
    display(chi2_results_df)

    print(
        "\nNumero feature categoriche originali significative p < 0.001:",
        chi2_results_df["significant_p005"].sum()
    )

else:
    chi2_results_df = pd.DataFrame(columns=[
        "feature",
        "test",
        "chi2_statistic",
        "p_value",
        "score_-log10_p",
        "significant_p005"
    ])

    print("\nNessuna feature categorica trovata per chi-square.")

# ============================================================
# GRAFICO UNICO FEATURE SIGNIFICATIVE:
# KRUSKAL-WALLIS + CHI-SQUARE SU FEATURE ORIGINALI
# ============================================================

kruskal_plot_df = kruskal_results_df.copy()
kruskal_plot_df = kruskal_plot_df[[
    "feature",
    "test",
    "p_value",
    "score_-log10_p",
    "significant_p005"
]]
kruskal_plot_df["feature_type"] = "Numerica"

chi2_plot_df = chi2_results_df.copy()
chi2_plot_df = chi2_plot_df[[
    "feature",
    "test",
    "p_value",
    "score_-log10_p",
    "significant_p005"
]]
chi2_plot_df["feature_type"] = "Categorica"

combined_stats_df = pd.concat(
    [kruskal_plot_df, chi2_plot_df],
    axis=0,
    ignore_index=True
)

combined_significant_df = combined_stats_df[
    combined_stats_df["significant_p005"]
].copy()

combined_significant_df = combined_significant_df.sort_values(
    by="p_value",
    ascending=True
)

top_n = 113
combined_top_plot = combined_significant_df.head(top_n).copy()

combined_top_plot = combined_top_plot.sort_values(
    by="score_-log10_p",
    ascending=True
)

fig, ax = plt.subplots(figsize=(max(13, 0.38 * len(combined_top_plot)), 7.8))


bars = ax.bar(
    combined_top_plot["feature"],
    combined_top_plot["score_-log10_p"],
    color=color_sequence(len(combined_top_plot), mode="gradient"),
    edgecolor="white",
    linewidth=0.7
)

polish_axes(
    ax,
    title="Top feature significative - Kruskal-Wallis e Chi-square",
    xlabel="Feature originale",
    ylabel="Score -log10(p-value)",
    grid_axis="y"
)


ax.axhline(
    -np.log10(P_VALUE_THRESHOLD),
    linestyle="--",
    linewidth=1.8,
    color=PLOT_COLORS["violet"],
    label="Soglia p = 0.001"
)


ax.invert_xaxis()


ax.tick_params(axis="x", labelrotation=90, labelsize=9)
for label in ax.get_xticklabels():
    label.set_horizontalalignment("center")
    label.set_verticalalignment("top")

ax.legend(frameon=True, facecolor="white", edgecolor="#DDE7F0", loc="upper right")
plt.tight_layout()
plt.show()

print("\nTop feature significative complessive:")
display(combined_top_plot.sort_values(by="p_value", ascending=True))


# Salvataggio tabelle statistiche
kruskal_file = "kruskal_preliminary_results.csv"
chi2_file = "chi2_preliminary_results.csv"
class_file = "class_distribution.csv"

kruskal_results_df.to_csv(kruskal_file, index=False, sep=";", decimal=",")
chi2_results_df.to_csv(chi2_file, index=False, sep=";", decimal=",")
class_distribution_df.to_csv(class_file, index=False, sep=";", decimal=",")

files.download(kruskal_file)
files.download(chi2_file)
files.download(class_file)


# ============================================================
# 11. SPLIT TRAIN/TEST FINALE 80/20
# ============================================================

target_col = "target_3class"
id_col = "hadm_id"

X = dataset_finale.drop(columns=[id_col, target_col])
y = dataset_finale[target_col]

label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

print("\nMapping classi:")
for idx, cls in enumerate(label_encoder.classes_):
    print(idx, "->", cls)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y_encoded
)

print("\nTrain shape:", X_train.shape)
print("Test shape:", X_test.shape)

print("\nDistribuzione train:")
print(pd.Series(y_train).value_counts(normalize=True).sort_index() * 100)

print("\nDistribuzione test:")
print(pd.Series(y_test).value_counts(normalize=True).sort_index() * 100)

numeric_cols, categorical_cols = get_feature_types(X_train)

print("\nNumero feature numeriche:", len(numeric_cols))
print("Numero feature categoriche:", len(categorical_cols))
print("Feature categoriche:", categorical_cols)


# ============================================================
# 12. NESTED STRATIFIED CV PER MODEL SELECTION
# ============================================================

print("\n" + "=" * 90)
print("NESTED STRATIFIED CROSS-VALIDATION PER MODEL SELECTION")
print("=" * 90)

outer_cv = StratifiedKFold(n_splits=OUTER_SPLITS, shuffle=True, random_state=RANDOM_STATE)
inner_cv = StratifiedKFold(n_splits=INNER_SPLITS, shuffle=True, random_state=RANDOM_STATE)

pipelines = build_pipelines(numeric_cols, categorical_cols, n_classes=len(label_encoder.classes_))

print("\nPipeline definite:")
for name in pipelines.keys():
    print("-", name)
print("Numero totale pipeline:", len(pipelines))

nested_results = []
nested_best_params = {}

for pipeline_name, item in pipelines.items():
    print("\n" + "=" * 90)
    print("Nested CV pipeline:", pipeline_name)
    print("=" * 90)

    outer_scores = []
    outer_best_params = []
    start_time_pipeline = time.time()

    for fold_idx, (train_idx, val_idx) in enumerate(outer_cv.split(X_train, y_train), start=1):
        X_outer_train = X_train.iloc[train_idx]
        X_outer_val = X_train.iloc[val_idx]
        y_outer_train = y_train[train_idx]
        y_outer_val = y_train[val_idx]

        grid = GridSearchCV(
            estimator=clone(item["pipeline"]),
            param_grid=item["params"],
            scoring="f1_macro",
            cv=inner_cv,
            n_jobs=-1,
            refit=True,
            return_train_score=False,
            verbose=0,
            error_score="raise"
        )

        grid.fit(X_outer_train, y_outer_train)
        y_outer_pred = grid.predict(X_outer_val)
        fold_macro_f1 = f1_score(y_outer_val, y_outer_pred, average="macro")

        outer_scores.append(fold_macro_f1)
        outer_best_params.append(grid.best_params_)

        print(f"Fold {fold_idx}: macro-F1 = {fold_macro_f1:.4f}, best params = {grid.best_params_}")

    elapsed_pipeline = time.time() - start_time_pipeline
    nested_best_params[pipeline_name] = outer_best_params

    nested_results.append({
        "pipeline": pipeline_name,
        "outer_macro_f1_mean": np.mean(outer_scores),
        "outer_macro_f1_std": np.std(outer_scores),
        "outer_scores": outer_scores,
        "best_params_by_outer_fold": outer_best_params,
        "time_sec": elapsed_pipeline
    })

nested_results_df = pd.DataFrame(nested_results).sort_values(
    by="outer_macro_f1_mean",
    ascending=False
)

print("\n=== RISULTATI NESTED CV ORDINATI ===")
display(nested_results_df)

nested_file = "nested_model_selection_results.csv"
nested_results_df.to_csv(nested_file, index=False, sep=";", decimal=",")
files.download(nested_file)

# Grafico risultati nested CV
plot_df = nested_results_df.sort_values(by="outer_macro_f1_mean", ascending=True)

fig, ax = plt.subplots(figsize=(12, 7.5))


ax.barh(
    plot_df["pipeline"],
    plot_df["outer_macro_f1_mean"],
    xerr=plot_df["outer_macro_f1_std"],
    color=color_sequence(len(plot_df)),
    edgecolor="white",
    linewidth=1.0,
    ecolor=PLOT_COLORS["dark_text"],
    capsize=4
)

polish_axes(
    ax,
    title="Nested Stratified CV - Confronto pipeline",
    xlabel="Macro-F1 media outer CV",
    ylabel="Pipeline",
    grid_axis="x"
)

plt.tight_layout()
plt.show()


# ============================================================
# 13. GRIDSEARCH FINALE UNA SOLA VOLTA SUL TRAINING 80%
#     Serve a ottenere gli iperparametri finali della pipeline scelta.
# ============================================================

best_pipeline_name = nested_results_df.iloc[0]["pipeline"]
print("\nMiglior pipeline da nested CV:", best_pipeline_name)

best_pipeline_config = pipelines[best_pipeline_name]

final_grid = GridSearchCV(
    estimator=clone(best_pipeline_config["pipeline"]),
    param_grid=best_pipeline_config["params"],
    scoring="f1_macro",
    cv=inner_cv,
    n_jobs=-1,
    refit=True,
    return_train_score=True,
    verbose=1,
    error_score="raise"
)

start_time = time.time()
final_grid.fit(X_train, y_train)
final_grid_time = time.time() - start_time

final_best_params = final_grid.best_params_
print("\nParametri finali scelti sul training 80%:")
print(final_best_params)
print("Macro-F1 media CV finale:", final_grid.best_score_)
print("Tempo GridSearch finale:", final_grid_time, "s")


# ============================================================
# 14. FUNZIONI PER STAMPARE FEATURE SELEZIONATE DAI FILTRI
# ============================================================

def print_selected_features_from_filter_pipeline(fitted_pipeline, pipeline_name, numeric_cols_used, categorical_cols_used):
    if "preprocessor" not in fitted_pipeline.named_steps:
        return

    preprocessor = fitted_pipeline.named_steps["preprocessor"]

    if not hasattr(preprocessor, "named_transformers_"):
        return

    print("\n" + "=" * 100)
    print(f"FEATURE SELEZIONATE - {pipeline_name}")
    print("=" * 100)

    # Numeriche
    if "num" in preprocessor.named_transformers_:
        try:
            num_pipe = preprocessor.named_transformers_["num"]
            num_selector = num_pipe.named_steps.get("selector", None)

            if num_selector is not None:
                num_support = num_selector.get_support()
                selected_num = np.array(numeric_cols_used)[num_support]
                num_scores = num_selector.scores_[num_support]
                num_pvalues = num_selector.pvalues_[num_support]

                num_df = pd.DataFrame({
                    "feature": selected_num,
                    "test": "Kruskal-Wallis",
                    "score_-log10_p": num_scores,
                    "p_value": num_pvalues
                }).sort_values(by="p_value", ascending=True)

                print("\nFeature numeriche selezionate con Kruskal-Wallis p < 0.001:")
                print("Numero:", len(num_df))
                display(num_df)
        except Exception as e:
            print("Errore feature numeriche:", e)

      # Categoriche
    if "cat" in preprocessor.named_transformers_:
        try:
            cat_pipe = preprocessor.named_transformers_["cat"]
            cat_selector = cat_pipe.named_steps.get("selector", None)

            if cat_selector is not None:
                cat_support = cat_selector.get_support()

                selected_cat = np.array(categorical_cols_used)[cat_support]
                cat_scores = cat_selector.scores_[cat_support]
                cat_pvalues = cat_selector.pvalues_[cat_support]

                cat_df = pd.DataFrame({
                    "feature": selected_cat,
                    "test": "Chi-square su feature originale",
                    "chi2_statistic": cat_scores,
                    "p_value": cat_pvalues
                }).sort_values(by="p_value", ascending=True)

                print("\nFeature categoriche originali selezionate con Chi-square p < 0.001:")
                print("Numero:", len(cat_df))
                display(cat_df)

        except Exception as e:
            print("Errore feature categoriche:", e)


if "filter" in best_pipeline_name:
    print_selected_features_from_filter_pipeline(
        fitted_pipeline=final_grid.best_estimator_,
        pipeline_name=best_pipeline_name,
        numeric_cols_used=numeric_cols,
        categorical_cols_used=categorical_cols
    )


# ============================================================
# 15. ABLATION STUDY SUL MIGLIOR MODELLO
#     Stessi iperparametri, cambia solo il gruppo di feature.
# ============================================================

print("\n" + "=" * 90)
print("ABLATION STUDY SUL MIGLIOR MODELLO")
print("=" * 90)
print("Miglior modello usato per ablation:", best_pipeline_name)
print("Iperparametri fissi:", final_best_params)

ablation_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
ablation_results = []

for group_name, group_cols in feature_groups.items():
    group_cols_existing = [col for col in group_cols if col in X_train.columns]

    X_train_group = X_train[group_cols_existing]
    group_numeric_cols, group_categorical_cols = get_feature_types(X_train_group)

    group_pipelines = build_pipelines(
        group_numeric_cols,
        group_categorical_cols,
        n_classes=len(label_encoder.classes_)
    )

    if best_pipeline_name not in group_pipelines:
        print("Pipeline non disponibile per gruppo:", group_name)
        continue

    estimator = clone(group_pipelines[best_pipeline_name]["pipeline"])
    estimator.set_params(**final_best_params)

    print("\nGruppo feature:", group_name, "- n feature:", len(group_cols_existing))

    scores = cross_val_score(
        estimator,
        X_train_group,
        y_train,
        cv=ablation_cv,
        scoring="f1_macro",
        n_jobs=-1
    )

    ablation_results.append({
        "feature_group": group_name,
        "n_features": len(group_cols_existing),
        "macro_f1_mean": np.mean(scores),
        "macro_f1_std": np.std(scores),
        "scores": scores
    })

    print("Macro-F1 media:", np.mean(scores))
    print("Std:", np.std(scores))

ablation_results_df = pd.DataFrame(ablation_results).sort_values(
    by="macro_f1_mean",
    ascending=False
)

print("\n=== RISULTATI ABLATION STUDY ===")
display(ablation_results_df)

ablation_file = "ablation_results.csv"
ablation_results_df.to_csv(ablation_file, index=False, sep=";", decimal=",")
files.download(ablation_file)

# Grafico ablation
plot_ablation = ablation_results_df.sort_values(by="macro_f1_mean", ascending=True)

fig, ax = plt.subplots(figsize=(10.5, 6.5))

# MODIFICA GRAFICO: ablation study colorato e leggibile.
ax.barh(
    plot_ablation["feature_group"],
    plot_ablation["macro_f1_mean"],
    xerr=plot_ablation["macro_f1_std"],
    color=color_sequence(len(plot_ablation)),
    edgecolor="white",
    linewidth=1.0,
    ecolor=PLOT_COLORS["dark_text"],
    capsize=4
)

polish_axes(
    ax,
    title=f"Ablation study - {best_pipeline_name}",
    xlabel="Macro-F1 media in CV",
    ylabel="Gruppo feature",
    grid_axis="x"
)

plt.tight_layout()
plt.show()

best_feature_group = ablation_results_df.iloc[0]["feature_group"]
print("\nMiglior gruppo feature da ablation:", best_feature_group)


# ============================================================
# 16. FIT FINALE E TEST FINALE UNA SOLA VOLTA
# ============================================================

final_group_cols = [col for col in feature_groups[best_feature_group] if col in X_train.columns]

X_train_final = X_train[final_group_cols]
X_test_final = X_test[final_group_cols]

final_numeric_cols, final_categorical_cols = get_feature_types(X_train_final)

final_pipelines_for_group = build_pipelines(
    final_numeric_cols,
    final_categorical_cols,
    n_classes=len(label_encoder.classes_)
)

final_model = clone(final_pipelines_for_group[best_pipeline_name]["pipeline"])
final_model.set_params(**final_best_params)

print("\nFit finale su tutto il training 80%...")
final_model.fit(X_train_final, y_train)

print("Valutazione finale sul test set 20%...")
y_pred = final_model.predict(X_test_final)

macro_f1 = f1_score(y_test, y_pred, average="macro")
weighted_f1 = f1_score(y_test, y_pred, average="weighted")
accuracy = accuracy_score(y_test, y_pred)
balanced_acc = balanced_accuracy_score(y_test, y_pred)
kappa = cohen_kappa_score(y_test, y_pred)

print("\n--- Performance test finale ---")
print("Pipeline finale:", best_pipeline_name)
print("Gruppo feature finale:", best_feature_group)
print("Macro-F1:", macro_f1)
print("Weighted-F1:", weighted_f1)
print("Accuracy:", accuracy)
print("Balanced accuracy:", balanced_acc)
print("Cohen kappa:", kappa)

print("\n--- Classification report ---")
print(classification_report(
    y_test,
    y_pred,
    target_names=label_encoder.classes_
))

# ============================================================
# CONFUSION MATRIX NORMALIZZATA
# ============================================================


cm_normalized = confusion_matrix(
    y_test,
    y_pred,
    normalize="true"
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_normalized,
    display_labels=label_encoder.classes_
)

fig, ax = plt.subplots(figsize=(6.5, 5.7))


disp.plot(
    ax=ax,
    values_format=".2f",
    cmap=CONFUSION_CMAP,
    colorbar=True
)

polish_axes(
    ax,
    title=f"Confusion Matrix normalizzata - {best_pipeline_name} - {best_feature_group}",
    xlabel="Classe predetta",
    ylabel="Classe reale",
    grid_axis="both"
)
ax.grid(False)
plt.tight_layout()
plt.show()

# Salvataggio risultati finali
test_results = pd.DataFrame([{
    "best_pipeline": best_pipeline_name,
    "best_feature_group": best_feature_group,
    "best_params": final_best_params,
    "macro_f1": macro_f1,
    "weighted_f1": weighted_f1,
    "accuracy": accuracy,
    "balanced_accuracy": balanced_acc,
    "cohen_kappa": kappa
}])

test_results_file = "test_final_results.csv"
test_results.to_csv(test_results_file, index=False, sep=";", decimal=",")
files.download(test_results_file)

print("\nFine pipeline.")
```
